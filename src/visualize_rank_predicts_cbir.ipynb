# CBIR Ranking Inspection

Load a fine-tuned CBIR checkpoint, select a catalogue from the configured dataset, and visualize its predicted neighbor order against the ground-truth ranks. Update `DATASET_PATH` and `MODEL_PATH` in the code cell for your environment and checkpoint layout before running.

```python
from cbir import visualize_predicted_ranking, load_model
from utils.catalogue_utils import build_catalogues
from models.image_encoders import GC_ViT_224
import torch
import random

DATASET_PATH = "/nas-ctm01/datasets/private/CINDERELLA/moe_organized_retrieval_CINDERELLA/image/pre_post"
MODEL_PATH = "/nas-ctm01/homes/dsviana/retrieval/checkpoints/cbir/best_model.pt"

device = "cuda" if torch.cuda.is_available() else "cpu"

model = GC_ViT_224()
model = load_model(model, MODEL_PATH, device=device)

catalogues = build_catalogues(DATASET_PATH)
catalogue = random.choice(catalogues)
print(f"Selected catalogue: {catalogue.catalogue_type}_{catalogue.catalogue_id}")

visualize_predicted_ranking(model=model, catalogue=catalogue, device=device)
```
